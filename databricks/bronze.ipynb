#Loading in data

##Init

```python
import datetime
current_date = datetime.datetime.now().strftime("%Y-%m-%d")
```

##Spotify songs

```python
df_spotify = spark.read.option("multiline", "true").json(f"s3://top-songs-bucket/raw/top_songs/spotify_top_songs/{current_date}/spotify_top_songs_{current_date}.json")
df_spotify.write.format("delta").mode("append").saveAsTable("top_songs_worldwide.bronze.spotify_songs")
```
